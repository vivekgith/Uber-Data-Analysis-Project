# Data Preprocessing

```python
dataset['PURPOSE'].fillna("NOT", inplace = True) #changing nan to NOT
```

```python
dataset.head()
```

```python
dataset['START_DATE'] = pd.to_datetime(dataset['START_DATE'], errors = 'coerce')# Change date time format syntex.\
dataset['END_DATE'] = pd.to_datetime(dataset['END_DATE'], errors = 'coerce')
```

```python
dataset.info() #Successfully changed datetime format.
```

```python
from datetime import datetime
dataset['date'] = pd.DatetimeIndex(dataset['START_DATE']).date # Make a new column for seprate date and time by this function. 
dataset['time'] = pd.DatetimeIndex(dataset['START_DATE']).hour
```

```python
dataset.head()
```

```python
dataset['day-night'] = pd.cut(x=dataset['time'],bins = [0,10,15,19,24],labels = ['Morning', 'Afternoon', 'Evening', 'Night']) #Making a categorization.
```

```python
dataset.head()
```

```python
dataset.dropna(inplace = True) # Removing Null values.
```

```python
dataset.shape
```

# Data Visualization

```python
plt.figure(figsize=(20,5))
plt.subplot(1,2,1)

sns.countplot(dataset['CATEGORY'],color='black')
plt.xticks(rotation = 90)

plt.subplot(1,2,2)
sns.countplot(dataset['PURPOSE'],color='black')
           
```

```python
sns.countplot(dataset['day-night'],color='black')
```

```python
dataset.head()
```

```python
dataset['MONTH'] =pd.DatetimeIndex(dataset['START_DATE']).month

month_label = {1.0:'Jan', 2.0:'Feb', 3.0:'Mar', 4.0:'April', 5.0:'May', 6.0:'June', 7.0:'July', 8.0:'Aug', 9.0:'Sep', 10.0:'Oct', 11.0:'Nov', 12.0:'Dec'}

dataset['MONTH'] = dataset.MONTH.map(month_label)

mon= dataset.MONTH.value_counts(sort=False)
```

```python
dataset.head()
```

```python
df = pd.DataFrame({
    "MONTHS":mon.values,
    "VALUE COUNT":dataset.groupby('MONTH',sort = False)['MILES'].max()
})

p = sns.lineplot(data = df)
p.set(xlabel="MONTHS", ylabel = "VALUE COUNT")                                 
```

```python
dataset['DAY'] = dataset.START_DATE.dt.weekday

day_label = {
    0:'Mon', 1:'Tue', 2:'Wed', 3:'Thu', 4:'Fri', 5:'Sat', 6:'Sun'}

dataset ['DAY'] = dataset['DAY'].map(day_label)
```

```python
dataset.head()
```

```python
day_label = dataset.DAY.value_counts()

sns.barplot(x=day_label.index, y= day_label,color='black')
plt.xlabel('DAY')
plt.ylabel('COUNT')
```
